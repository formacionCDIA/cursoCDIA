#### Gráfico de línea

El gráfico más básico es el gráfico de línea. Los gráficos de líneas se utilizan para mostrar el valor cuantitativo en un intervalo o intervalo de tiempo continuo. Se usa con mayor frecuencia para mostrar tendencias y relaciones (cuando se agrupan con otras líneas).

Para crear este gráfico llamar a la función `plot()` y pasarle los argumentos `x` e `y` que tienen que ser 2 listas o arrays del mismo tamaño.


```python
# Datos
# array - lista de números entre el valor 0 y 5, con un incremento de 0,1 entre los valores
x = np.arange(0,5, 0.1)
print('X:',x)
# Función seno para la lista de números
y = np.sin(x)
print('y:',y)
```

```python
fig = plt.figure() # Para crear gráfica
ax = plt.axes()
ax.plot(x,y)
plt.show()
```

Además de los datos `x` e `y`, se pueden indicar otros argumentos a la función `plot` para personalizar este tipo de gráfica. Cómo por ejemplo:

```python
# Para crear gráfica personalizada
fig = plt.figure()
ax = plt.axes()
# Cada tipo de grafica puede ser personalizada, mirar su documentación
ax.plot(x, y, color='green', linewidth=5, linestyle='--')
plt.show()
```

También podemos (y siempre deberíamos) agregar títulos a los gráficos y agregar etíquetas (labels) a los ejes para ser lo más claros y explícitos posibles.


```python
# Para crear gráfica personalizada
fig = plt.figure()
ax = plt.axes()
ax.plot(x, y, color='green', linewidth=5, linestyle='--')

# Titulo y etiquetas para los ejes simples
ax.set_title("seno(x) entre 0 y 5")
ax.set_xlabel('etiqueta x')
ax.set_ylabel('etiqueta Y')
plt.show()
```

Pero podriamos también, generar una grilla de ejes simples, tal que en cada celda crear distintos tipos de gráficos para distintos datos. Para ello se usa la función `subplots()`.

**Caso 1**: misma figura, mismos ejes simples, más de un tipo de gráfico.

```python
x2 = np.linspace(0, 5) # Datos
y2 = x2 / 5
```

```python
# Crear 2 gráficas en misma figura utilizando los mismos ejes
fig, ax = plt.subplots()

# Crear 1er gráfico de línea en eje simple (1,1)
ax.plot(x, y, color='green', linewidth=5, linestyle='--', label='Seno')
# Crear 2do gráfico de línea en eje simple (1,1)
ax.plot(x2, y2, color='red', linewidth=5, linestyle='-', label='recta')

ax.set_title("seno(x) entre 0 y 5")
ax.set_xlabel('etiqueta x')
ax.set_ylabel('etiqueta Y')

# Para mostrar leyenda
ax.legend()
plt.show()
```

**Caso 2**: misma figura, grilla de ejes simples, más de un tipo de gráfico y distintos datos.

```python
# Crear gráfica con plt.subplots
# Dos variables fig y axes
fig, axes = plt.subplots(nrows=2, ncols=1) # figura y grilla con 2 fias y 1 columna. Ver como una lista.
# Título a los subplots
fig.suptitle("Titulo subplots")

# Personalizar cada ejes simples de la grilla
axes[0].plot(x, y, color='green', linewidth=5, linestyle='--', label='Seno')
axes[1].plot(x2, y2, color='red', linewidth=5, linestyle='-', label='recta')

# Para mostrar leyenda
axes[0].legend()
axes[1].legend()
plt.show()
```

```python
# Otra forma utilizando plt.subplot
fig = plt.figure()
fig.suptitle("Titulo subplots")

# Tres enteros (fila, columna, orden).
# La subtrama ocupará la posición de index en una cuadrícula con nrows filas y ncols columnas.
plt.subplot(2,1,1)
plt.plot(x, y, color='green', linewidth=5, linestyle='--', label='Seno')
plt.legend()
plt.subplot(2,1,2)
plt.plot(x2, y2, color='red', linewidth=5, linestyle='-', label='recta')
plt.legend()
plt.show()
```

#### Histograma

Cuando queremos visualizar la distribución de una variable numérica es muy útil utilizar histogramas. Los histogramas resumen visualmente la distribución de una variable numérica continua midiendo la frecuencia con la que determinados valores aparecen en el conjunto de datos.

En un histograma, el eje x es una línea numérica que se ha dividido en rangos de números o bins. Para cada bin, se dibuja una barra en la que el ancho de la barra representa el rango del bin y la altura de la barra representa el número de puntos de datos incluidos en ese rango.

Para crear está gráfica, llamar la función `hist()`. En cuanto argumentos `x`
puede ser solo un array o una secuencia de arrays que no es necesario que tengan la misma longitud. Veamos el siguiente ejemplo:

```python
# Cantidad de valores
N_valores = 100
# Crear un generador de números aleatorios con una semilla fija para la reproducibilidad
rng = np.random.default_rng(1)
# Distribución normal, media 0 y desvio estándar 1
x = rng.standard_normal(N_valores)
x
```

```python
# Crear Histograma
fig, ax = plt.subplots()
# Titulo
fig.suptitle("Histograma")
ax.hist(x, bins='auto') # defecto bins = 10

# Personalizar más
ax.set_xlabel('bins')
ax.set_ylabel('Frecuencia')
plt.show()
```

¿Qué pasa si usamos un valor muy grande en bins?

```python
# Crear Histograma
fig, ax = plt.subplots()
# Titulo
fig.suptitle("Histograma")
ax.hist(x, bins=200)

# Personalizar más
ax.set_xlabel('bins')
ax.set_ylabel('Frecuencia')
plt.show()
```

¿Y muy chico?

```python
# Crear Histograma
fig, ax = plt.subplots()
# Titulo
fig.suptitle("Histograma")
ax.hist(x, bins=4)


# Personalizar más
ax.set_xlabel('bins')
ax.set_ylabel('Frecuencia')
plt.show()
```

#### Gráfico de Barra
El gráfico de barras clásico utiliza barras horizontales o verticales (gráfico de columnas) para mostrar comparaciones numéricas discretas entre categorías. Un eje del gráfico muestra las categorías específicas que se comparan y el otro eje representa una escala de valores discretos y por lo tanto responden a la pregunta de ¿cuántos? en cada categoría.

Para crear esta gráfica, llamar la función `bar()`. Las barras se posicionan en `x` con la alineación dada, barras verticales. Sus dimensiones se dan por altura (`height`) y ancho (`width`).

```python
# Datos
x = ['Manzana', 'Pera', 'Mango', 'Naranja'] # para eje x (categoría)
y = [40, 100, 30, 55] # para eje y (frecuencia)
```

```python
# Crear gráfico de barra
fig, ax = plt.subplots()
fig.suptitle("Gráfico de barra")

ax.bar(x=x, height=y)

ax.set_xlabel('Categorias')
ax.set_ylabel('Frecuencia')
plt.show()
```

#### Diagrama de dispersión

Los diagramas de dispersión usan una colección de puntos dispuestos en coordenadas cartesianas para mostrar valores de dos variables númericas. Al mostrar una variable en cada eje, se puede detectar si existe una relación o correlación entre las dos variables.
Los diagramas de dispersión son ideales cuando se tienen datos numéricos emparejados y se desea ver si una variable afecta a la otra. Sin embargo, recuerde que la correlación no es causal y otra variable inadvertida puede estar influyendo en los resultados.

Para crear está gráfica, llamar la función `scatter()`. En cuanto argumentos `x` ,`y` puede ser un escalar o un array o semejante.



```python
import pandas as pd

# Datos
n = 1000 # Cantidad de valores

# loc=0.0 (media), scale=1.0 (desvio estandar)
x = np.random.normal(size = n)
y = np.random.normal(size = n)
```

```python
# Crear gráfico de barra
fig, ax = plt.subplots()
fig.suptitle("Diagrama de dispersión")

ax.scatter(x, y)
ax.set_xlabel('valores de x')
ax.set_ylabel('valores de y')
plt.show()
```

```python
n = 100
# Datos
x = np.linspace(-1, 1, n) + 0.25 * np.random.normal(size = n)
y = 3 * x + 0.25 * np.random.normal(size = n)

# Crear gráfica
fig, ax = plt.subplots()
fig.suptitle("Diagrama de dispersión")

ax.scatter(x, y)
ax.set_xlabel('valores de x')
ax.set_ylabel('valores de y')
plt.show()
```

#### Tipos de funciones en Seaborn : Figure-level vs Axes-level
* **Axes-level:** grafican los datos en un objeto `matplotlib.pyplot.Axes`, que es el valor de retorno de la función. Teniendo en cuenta el diagrama anterior serían las funciones dentro de los grandes módulos, ejemplo `histplot()`, `lineplot()`,
 y `barplot()`. La creación de gráficos a nivel de ejes es similar a la creación de gráficos de matplotlib.
* **Figure-level:** grafican los datos en un objeto `FacetGrid`. Serían por ejemplo las funciones. `replot()`, `displot()`y `catplot()`.

#### Gráfico de línea

Esto se logra con la función `lineplot()`, ya sea directamente o con `relplot()` configurando `kind="line"`.

En Seaborn los argumentos de la función que indican los datos a graficar son  `Data` o  `x`,`y`.

* `Data` es la estructura de datos de entrada, que puede ser `pandas.DataFrame` o `numpy.ndarray`. Esta es una de las diferencias con Matplotlib.
* `x` e  `y`, especifican las posiciones en los ejes _x_ e _y_, son vectores o las claves en los datos, como los nombres de las columnas del `DataFrame`.  

Vamos a trabajar con el siguiente dataset:

```python
# Conjunto de datos sobre una Aerolínea
# Selecionar datos
flights = sns.load_dataset("flights")
flights.head()

# Mostrar

flights.head(10)
```

##### Crear gráfico por una función Axes-level

```python
# Crear gráfico de linea por una función Axes-level
# Por defecto estimator='mean'. Método para agregar múltiples observaciones de la variable 'y' en el mismo nivel 'x'.
# Si se selecciona Ninguno, se extraerán todas las observaciones.
# errorbar=(2 items) ('ci', 95) - Nombre del método errorbar

#Cuando se muestra una estadística de resumen, normalmente es adecuado agregar barras de error, que brindan una pista visual sobre qué tan bien el resumen representa los puntos de datos subyacentes.
sns.lineplot(data=flights, x="year", y="passengers")
```

En base al tipo de gráfica, gráfico de línea, podriamos indicar los otros argumentos de la función:

```python
sns.lineplot(data=flights, x="year", y="passengers", errorbar=None)
```

Asignando una semántica de agrupación como `hue` (tono),`size`(tamaño) o `style` (estílo) para trazar líneas separadas.

```python
sns.lineplot(data=flights, x="year", y="passengers", hue='month')
```

```python
sns.lineplot(data=flights, x="year", y="passengers", hue="month", style="month")
```

Si cambiamos la organización del `DataFrame` utilizando la función `pivot()` , para que la columna 'year' sea el índice del `DataFrame` y los valores de la columna 'month' sean las columnas.

```python
# Re-organizar
flights_wide = flights.pivot(index="year", columns="month", values="passengers")

# Mostrar
flights_wide.head()
```

Podríamos llamar a la función `sns.lineplot` solo indicando la estructura de datos.

```python
sns.lineplot(data=flights_wide)
```

Cada línea en la figura grafica las columnas de mes. Similiar a la gráfica previamente generada en donde otra era la entrada de datos para la cual también fue necesario indicar los argumentos `hue` y  `style` en la función `lineplot()` .  

Para personalizar una figura de Seaborn, agregar títulos, etiquetas, tamaño, entre otros. Existe varias formas:

**Por parámetros explícitos de la función Seaborn**

`matplotlib.axes.Axes.set()`

```python
#  función Axes-level devuelve un objeto matplotlib.axes.Axes
# Recordar los objetos de  matplotlib y los argumentos asociados para el tipo de gráficos
g = sns.lineplot(data=flights_wide)
g.figure.suptitle('Cantidad de pasajeros entre los años 1949 a 1960')
g.figure.set_size_inches((10, 4))
g.set(
    xlabel="Años",
    ylabel="Cantidad de Pasajeros",

)
g
```

**Por las funciones de Matplotlib**

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 1, figsize=(10, 4))
fig.suptitle('Cantidad de pasajeros entre los años 1949 a 1960')

sns.lineplot(data=flights_wide, ax=ax)

ax.set_xlabel("Años")
ax.set_ylabel("Cantidad de Pasajeros")
ax.legend(loc= "upper left")

```

### Seaborn y Matplotlib

* Matplotlib es una biblioteca proporciona una amplia gama de herramientas para crear visualizaciones altamente personalizables. Seaborn está construida encima de ella. Muchas acciones pueden realizarse solo con las funciones de Seaborn. Otras opciones de personalización pueden requerir el uso directo de Matplotlib.

* Seaborn, proporciona una interfaz más sencilla e intuitiva para crear gráficos estadísticos comunes. Seaborn está diseñado para trabajar con `DataFrame` de Pandas, lo que facilita la creación de visualizaciones con un código mínimo. También ofrece una serie de funciones estadísticas integradas, que permiten a los usuarios realizar fácilmente complejos análisis estadísticos con sus visualizaciones.

* Seaborn, por su parte, ofrece una gama de estilos y paletas de colores predeterminados optimizados para distintos tipos de datos y visualizaciones. Esto facilita a los usuarios la creación de gráficos visualmente atractivos con una personalización mínima.


```python
# Crear histograma con matplotlib
fig, axs = plt.subplots(1,2,figsize=(10, 4), sharey=True)
fig.suptitle("Histograma")
axs[0].hist(tips.loc[tips['sex']=='Male','total_bill'],color='red',alpha=0.7,  bins=4)
axs[1].hist(tips.loc[tips['sex']=='Female','total_bill'],color='blue',alpha=0.7, bins=4)
axs[0].set_xlabel('bins')
axs[0].set_ylabel('Frecuencia')
axs[1].set_xlabel('bins')
plt.show()
```

```python
# # Crear Histograma con seaborn
g = sns.displot(tips, x="total_bill",  bins=4,  palette=['red','blue'], kind='hist', hue='sex',  col='sex')
g.figure.suptitle('Histograma')
g.figure.set_size_inches((10, 4))
g.set( xlabel='bins',ylabel='Frecuencia')
g
```

#### Paleta de colores Seaborn

El color desempeña un papel muy importante en la visualización de datos. El color añade varias dimensiones a un gráfico cuando se utiliza eficazmente.
Una paleta es una superficie plana en la que un pintor mezcla pinturas.

Seaborn proporciona una función llamada `color_palette()`. Puede utilizarse para dar color a los gráficos y añadirles valor estético. Devuelve una lista de colores que definen una paleta de colores.

Hay varias paletas Seaborn disponibles. Estas son:

- Deep

- Muted

- Bright

- Pastel

- Dark

- Colorblind

Además de estas también podemos crear nuevas paletas.

Hay otra función `seaborn.palplot()` que se ocupa de las paletas de colores. Esta función traza la paleta de colores como una matriz horizontal.

[Información adicional](https://seaborn.pydata.org/tutorial/color_palettes.html)

**Paletas de colores cualitativas**

Las paletas cualitativas o categóricas son las más adecuadas para graficar datos categóricos.

```python
# Default
import matplotlib.pyplot as plt


current_palette1 = sns.color_palette()

sns.palplot(current_palette1)

plt.show()
```

**Paletas de colores secuenciales**

Los graficos secuenciales son adecuados para expresar la distribución de datos que van de valores relativos más bajos a valores más altos dentro de un rango. Añadiendo un carácter adicional "s" al color pasado al parámetro color se genera el gráfico Secuencial.

Por ejemplo: tenemos que añadir "s" al parámetro como "Greens" de la siguiente manera:

```python
current_palette2 = sns.color_palette("Blues")

sns.palplot(current_palette2)

plt.show()
```

**Paleta de colores divergentes**

Las paletas divergentes utilizan dos colores diferentes. Cada color representa la variación en el valor que va desde un punto común en cualquier dirección.

Suponemos que se trazan los datos que van de -1 a 1. Los valores de -1 a 0 toman un color y de 0 a +1 toman otro color.

Por defecto, los valores se centran a partir de cero. Podemos controlarlo con el parámetro center pasando un valor como sigue:


```python
current_palette3 = sns.color_palette()

sns.palplot(sns.color_palette("BrBG", 10))

plt.show()
```

#### Gráfico de barras para variables categóricas

* Propósito: comparar frecuencias o promedios por categoría.

```python
# Dataset de ejemplo
df = sns.load_dataset("titanic")

# Gráfico de barras: proporción de sobrevivientes por clase
# interfaz orientada objeto

fig, ax = plt.subplots(figsize=(8, 5))
sns.countplot(data=df, x="class", hue="survived", palette="Set2", ax=ax)
ax.set_title("Supervivencia por clase en el Titanic")
ax.set_xlabel("Clase")
ax.set_ylabel("Cantidad de personas")
ax.legend(title="Sobrevivió", labels=["No", "Sí"])
plt.tight_layout()
plt.show()
```

#### Boxplot para variables numéricas y categóricas
* Propósito: visualizar distribución, mediana y posibles outliers.

```python
fig, ax = plt.subplots(figsize=(8, 5))
sns.boxplot(data=df, x="class", y="age", palette="pastel", hue='class', legend=False, ax=ax)
ax.set_title("Distribución de edades por clase")
ax.set_xlabel("Clase")
ax.set_ylabel("Edad")
plt.tight_layout()
plt.show()
```

#### Histogramas para variables numéricas
* Propósito: ver distribución de una variable numérica.

```python
fig, ax = plt.subplots(figsize=(8, 5))
sns.histplot(data=df, x="age", bins=30, kde=True, color="Blue", ax=ax)
ax.set_title("Distribución de edad")
ax.set_xlabel("Edad")
ax.set_ylabel("Frecuencia")
plt.tight_layout()
plt.show()
```

#### Gráfico de dispersión con variables numéricas
* Propósito: analizar relación entre dos variables numéricas.

```python
df_p = sns.load_dataset("penguins").dropna()

fig, ax = plt.subplots(figsize=(8, 5))
sns.scatterplot(data=df_p, x="bill_length_mm", y="flipper_length_mm",
                hue="species", palette="Dark2", ax=ax)
ax.set_title("Relación entre longitud del pico y de la aleta")
ax.set_xlabel("Longitud del pico (mm)")
ax.set_ylabel("Longitud de la aleta (mm)")
ax.legend(title="Especie")
plt.tight_layout()
plt.show()
```

#### Heatmap para correlación entre variables numéricas
* Propósito: detectar relaciones lineales.

```python
fig, ax = plt.subplots(figsize=(8, 6))
corr = df_p.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f", ax=ax)
ax.set_title("Mapa de calor de correlaciones")
plt.tight_layout()
plt.show()
```

#### Ejemplo de dos visualizaciones que buscan responder ¿Cómo varía el nivel de satisfacción con el trabajo según la edad?"

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Simulamos un dataset
np.random.seed(42)
df = pd.DataFrame({
    "Edad": np.random.randint(18, 65, 200),
    "Satisfacción": np.random.choice(["Baja", "Media", "Alta"], 200)
})

# Mala visualización
fig, ax = plt.subplots(figsize=(8, 5))
sns.countplot(data=df, x="Edad", hue="Satisfacción", ax=ax)  # Demasiadas barras
ax.set_title("Resultados")
ax.set_xlabel("Edad")
ax.set_ylabel("Cantidad")
plt.tight_layout()
plt.show()
```

* Demasiadas categorías en el eje x (Edad - continuo), se debería agrupar en rangos.

* El título es vago (“Resultados”).

* No se eligió una paleta clara.

* No hay guía visual que ayude a interpretar.

```python
# Agrupar por rangos de edad
df["Grupo_Edad"] = pd.cut(df["Edad"], bins=[18, 30, 45, 60, 70],
                          labels=["18-30", "31-45", "46-60", "61-70"])

# Paleta default tipo arcoiris (no apta para daltónicos)
fig, ax = plt.subplots(figsize=(8, 5))
sns.countplot(data=df, x="Grupo_Edad", hue="Satisfacción",
              palette=["red", "green", "blue"], ax=ax)

ax.set_title("Nivel de satisfacción laboral según grupo etario", fontsize=14)

ax.set_xlabel("Grupo de edad")
ax.set_ylabel("Cantidad de personas")
ax.legend(title="Satisfacción laboral")
plt.tight_layout()
plt.show()
```
