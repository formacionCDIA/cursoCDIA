## 1. Objetivos y alcances

**Contexto**

Trabajas en una consultora de ciencias de datos. El cliente que solicita ayuda corresponde a una Bodega X con vinos en Argentina y Chile. Desea optimizar los precios de sus vinos por lo que considera importante analizar el mercado. Dispone de datos externos sobre reseñas de vinos.

**Objetivo general**

Optimizar el precio de vino de Argentina y Chile de la Bodega X.

**Conjunto de datos**




* Los datos fueron escrapeados del sitio WineEnthusiast en el año 2017. Estan almacenados en un archivo csv, `winemag-data-130k-v2.csv`.

Para más información clicar en [Reseña de vino en Kaggle](https://www.kaggle.com/zynicide/wine-reviews/)


Descripción del conjunto de datos:

* country: El país de origen del vino.
* description: Descripción del vino
* designation: El viñedo dentro de la bodega de donde provienen las uvas que hicieron el vino.
* points: La cantidad de puntos que WineEnthusiast calificó al vino en una escala de 1 a 100 (aunque dicen que solo publican reseñas de vinos que puntúan &gt;=80).
* price: El costo de una botella de vino.
* province: La provincia o estado de donde proviene el vino.
* region_1: La zona de cultivo de vino en una provincia o estado (por ejemplo, Napa).
* region_2: A veces, hay regiones más específicas especificadas dentro de una zona de cultivo de vino (por ejemplo, Rutherford dentro del Valle de Napa), pero este valor a veces puede estar en blanco.
* taster_name: nombre del catador.
* taster_twitter_handle: usuario de twitter del catador.
* title: El título de la reseña del vino.
* variedad: El tipo de uva utilizada para hacer el vino (por ejemplo, Pinot Noir).
* winery: La bodega que elaboró ​​el vino.

Ahora Seleccionemos todos los vinos cuyo origen de país sean de Argentina y Chile. Sabemos que los valores en 'country' contienen a 'Argentina' y 'Chile'.

```python
 # Asignar  el subconjunto de vinos de los paises definidos
mascara = reseVino_df['country'].isin(['Argentina','Chile']) # Condición
pais_vino_df = reseVino_df[ mascara ]


pais_vino_df.head()
```

De manera similar a la expresión condicional, la función condicional `isin` devuelve un valor verdadero para cada fila en la que se encuentran los valores de la lista proporcionada. Para filtrar las filas en función de dicha función, se utiliza la función condicional dentro de los corchetes de selección [].



```python
pais_vino_df.shape # Dímensión
```

```python
pais_vino_df.info() # Resumen técnico
```

Se observa que hay menos variables con datos faltantes
